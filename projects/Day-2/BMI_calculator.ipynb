```python
def bmi_calculator(weight, height):
    
    bmi = weight / (height ** 2)
    
    if bmi<18.5:
        res='underweight'
    elif bmi >=18.5 and bmi<25:
        res='normal weight'
    elif bmi>=25:
        res='overweight'
    return res

weight = float(input('enter your weight in kg= '))
height = float(input('enter your height in m= '))
res=bmi_calculator(weight, height)
print('You are', res)
```
