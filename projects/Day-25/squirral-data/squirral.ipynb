```python
import pandas as pd

df=pd.read_csv('2018_Central_Park_Squirrel_Census_-_Squirrel_Data.csv')

df.head()
```

```python
df['Primary Fur Color'].unique()
```

```python
gray_sq_count=len(df[df['Primary Fur Color']=='Gray'])
print(f'number of gray squirrals= {gray_sq_count}')
```

```python
black_sq_count=len(df.loc[df['Primary Fur Color']=='Black'])
print(f'number of black squirrals= {black_sq_count}')
```

```python
cinn_sq_count=len(df.loc[df['Primary Fur Color']=='Cinnamon'])
print(f'number of cinnamon squirrals= {cinn_sq_count}')
```

```python
data_dict={
    'Fur color':['Gray','Cinnamon','Black'],
    'Number':[gray_sq_count,cinn_sq_count,black_sq_count]
}

color_df=pd.DataFrame(data_dict)
color_df.to_csv('Squirral_color_counts.csv')
```
